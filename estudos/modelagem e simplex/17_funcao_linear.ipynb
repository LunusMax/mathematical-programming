# Exercício 17 — Otimização linear

Considere o problema

$$
\min \; f(x_1,x_2) = -x_1 - x_2
$$

sujeito a

$$
\begin{aligned}
x_1 + x_2 &\leq 6,\\
x_1 - x_2 &\leq 4,\\
x_2 &\leq 1,\\
x_1,x_2 &\geq 0.
\end{aligned}
$$

```python
import sympy as sp
from sympy.plotting import plot_implicit
```

```python
x1, x2 = sp.symbols('x_1 x_2', real=True)
```

```python
x3, x4, x5 = sp.symbols('x_3 x_4 x_5', real=True)
```

```python
restricoes = sp.And(
    x1 + x2 <= 6,
    x1 - x2 <= 4,
    x2 <= 1,
    x1 >= 0,
    x2 >= 0
)
```

```python
p = plot_implicit(
    restricoes,
    (x1, -1, 7),
    (x2, -1, 3),
    show=False
)
```

```python
retas = [
    sp.Eq(x1 + x2, 6),
    sp.Eq(x1 - x2, 4),
    sp.Eq(x2, 1),
    sp.Eq(x1, 0),
    sp.Eq(x2, 0)
]

for reta in retas:
    grafico_reta = plot_implicit(
        reta,
        (x1, -1, 7),
        (x2, -1, 3),
        show=False
    )

    p.extend(grafico_reta)

p.show()
```
